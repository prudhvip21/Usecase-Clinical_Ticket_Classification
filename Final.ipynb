## Text Preprocessing

*As detailed above one of the key issue for text classification problem posed here is preprocessing of `DATA` column. It is understood to be rtf format. While libraries can be used to convert, we have used regular expressions. First the data is split into words. After splitting each word is matched for the format. There were some repeating patterns like '\\par\' and \\x080' . Such types were included in the pattern to be removed by regular expressions.  After having a manual look at data, some of the words were repeating. So they were added to stop words.*

```python
stop = stopwords.words('english')
rm_list = ["phonexxxx-xxxx","for:","additional","Phone","Note", "call", "patient", "back", "at:", "cell", "phone", "caller", "from", "patient", "caller:", "name:", "other", "taken:", "by:", "Follow-up","Details:","default" ,"Action","Taken","Default","#","%","&","'","''","'*","'+","'-","'..","'-thomas","'..next","'.please","'/","'="]
stop.extend(rm_list)
```

```python
full_data = list(med_data['DATA']) # Copying the whole data into a list
```

*Once all the stop words are defined we do the following:*

- Iterate over each data point
- For each data point iterate over each tokens 
- For each token we match for all white space charecters and other formats which are generally found in data. If any match is found, the loop removes it. All words which doesn’t match the above expression are added to list of corpus `final_data`.
- And at the very end we remove all the stop words from the `final_data` 

```python
final_data= list()
for data in full_data:
    split = word_tokenize(data)
    regex = re.compile('\s+')
    new = []
    tokens =[]
    for i in range(len(split)) :
        if len(regex.findall(split[i]))==0 and(re.search(r'\\|\x080|\d|xxxx-xxxx| |//|#|&|\|,' ,split[i], re.M|re.I))==None :
            new.append(str(split[i].lower()))
    tokens = [token for token in new if token not in stop]
    final_data.append(str(list(tokens)))
```

**re.I:**<br> *Perform case-insensitive matching; expressions like [A-Z] will also match lowercase letters.*

**re.M**<br> *When specified, the pattern character `^` matches at the beginning of the string and at the beginning of each line (immediately following each newline); and the pattern character `$` matches at the end of the string and at the end of each line (immediately preceding each newline). By default, `^` matches only at the beginning of the string, and `$` only at the end of the string and immediately before the newline (if any) at the end of the string.*

*If a word in final_names doesn't have a synset than it is considered a mistake.*

```python
spells = [token for token in final_names if len(wordnet.synsets(token)) > 0 ] 
spells = list(set(spells))
mistakes = [token for token in final_names if len(wordnet.synsets(token)) == 0 ] 
```

*Now for all the mistakes, we get word from our corpus that are 1 Edit distance or 2 Edit Distance away from the mistaken word.*

```python
def words(text): return re.findall(r'\w+', text.lower())

WORDS = Counter(spells)

def P(word, N=sum(WORDS.values())): 
    "Probability of `word`."
    return WORDS[word] / N

def correction(word): 
    "Most probable spelling correction for word."
    return max(candidates(word), key=P)

def candidates(word): 
    "Generate possible spelling corrections for word."
    return (known([word]) or known(edits1(word)) or known(edits2(word)) or [word])

def known(words): 
    "The subset of `words` that appear in the dictionary of WORDS."
    return set(w for w in words if w in WORDS)

def edits1(word):
    "All edits that are one edit away from `word`."
    letters    = 'abcdefghijklmnopqrstuvwxyz'
    splits     = [(word[:i], word[i:])    for i in range(len(word) + 1)]
    deletes    = [L + R[1:]               for L, R in splits if R]
    transposes = [L + R[1] + R[0] + R[2:] for L, R in splits if len(R)>1]
    replaces   = [L + c + R[1:]           for L, R in splits if R for c in letters]
    inserts    = [L + c + R               for L, R in splits for c in letters]
    return set(deletes + transposes + replaces + inserts)

def edits2(word): 
    "All edits that are two edits away from `word`."
    return (e2 for e1 in edits1(word) for e2 in edits1(e1))
```

*Iterate over each word correct the spells.*

```python
d = {}

for word in mistakes:
    k = correction(word)
    if k != word : 
        d[word] = k

spell_cor = final_data 
```

*After correcting the mistakes, we make change to our final_data for the misspelled words.*

```python
for lst in range(len(spell_cor)):
    for j in range(len(spell_cor[lst])) :
        if spell_cor[lst][j] in d.keys():
            spell_cor[lst][j] = d[spell_cor[lst][j]]   
```
